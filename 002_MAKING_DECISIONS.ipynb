```python
# 1. DMOJ problem ccc06j1, Canadian Calorie Counting
calories = 0
order = []
while True:
  print("1. burger")
  print("2. drinks")
  print("3. side item")
  print("4. dessert")

  choice = input("enter your choice:")

  if choice == '1':
    ask = input(
        "1 – Cheeseburger (461 Calories)"
        "2 – Fish Burger (431 Calories)"
        "3 – Veggie Burger (420 Calories)"
        "4 – no burger"

    )
    if ask == "1":
      calories += 461
      order.append('cheeseburger')
      print("cheeseburger added")
    elif ask == '2':
      calories += 431
      order.append('fish burger')
      print("fish burger added")
    elif ask == '3':
      calories += 420
      order.append('veggie burger')
    elif ask == '4':
      calories += 0
      order.append('no burger')
    else:
      print("invalid input")





```

```python
# ccc15j1
from datetime import date

month = int(input())
day = int(input())

given_date = date(2015, month, day)
special_date = date(2015, 2, 18)

if given_date < special_date:
    print("Before")
elif given_date == special_date:
    print("Special")
else:
    print("After")
```
