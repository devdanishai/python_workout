```python
# 12. Write a function to return True if the first and last number of a given list is the same. If the numbers are different, return False.

x = [10, 20, 30, 40, 10]
y = [75, 65, 35, 75, 30]
def checker(num):
  if num[0] == num[-1]:
    return True
  else:
    return False

checker(y)
checker(x)
```

```python
# 16 Write a program to check if a given number is a palindrome (reads the same forwards and backwards).

def palin(num):
  num = num
  onum = num
  rnum = 0

  while num > 0:
    digit = num % 10
    rnum = rnum * 10 + digit
    num = num // 10

  if onum == rnum:
    return "palindrome"
  else:
    return "not palindrome"

d = 123
print(palin(d))
```
