````python
# 1. Reverse a Number — Input 76589 → Output 98567
num = 76589
r_num = 0

while num > 0:
  digit = num % 10
  r_num = r_num * 10 + digit
  num = num // 10
print(r_num)
````

````python
#2. Sum of Digits — Input 1234 → Output 10

num = 1234
num1 = []

while num > 0:
  digit = num % 10
  num1.append(digit)
  num = num // 10

print(num1)
sum1 = 0
for i in num1:
  sum1 += i
print(sum1)


````

````python
# 3. Count of Digits — Input 98765 → Output 5
num = 98765
counter = 0
while num > 0:
  digit = num % 10
  counter += 1
  num = num // 10
print(counter)
````

````python
# 4. Product of Digits — Input 1234 → Output 24
num = 1234
product = 1
while num > 0:
  digit = num % 10
  product *= digit
  num = num //10
print(product)
````

````python
# 5. Find Maximum Digit — Input 4271 → Output 7
num = 4271
largest=0
while num> 0:
  digit = num % 10
  if digit > largest:
    largest = digit
  num = num //10
print(largest)
````

````python
# 6. Find Minimum Digit — Input 4271 → Output 1
num = 4271
smallest = 9

while num> 0:
  digit = num % 10
  if digit < smallest:
    smallest = digit
  num = num // 10
print(smallest)
````

````python
# 7. Palindrome Number Check — 121 → True, 123 → False

num = 123
o_num = num
r_num = 0

while num > 0:
  digit = num % 10
  r_num = r_num * 10 + digit
  num = num // 10

if o_num == r_num:
  print(f"{o_num} is palindrome number")
else:
  print(f"{o_num} is not palindrome number")
````

````python
# 9. Armstrong Number (Narcissistic Number) — 153 → True (1³+5³+3³ = 153)
num = 370
o_num = num
res = 0
while num > 0:
  digit = num % 10
  res = res + digit ** 3
  num = num //10
if res == o_num:
  print("armstrong number")
else:
  print(" not armstrong number")
````

````python
# 10. Count Occurrences of a Given Digit — Find how many times digit 7 appears in 9727257
num = 9727257
target = 7
counter = 0
while num > 0:
  digit = num % 10
  if digit == target:
    counter += 1
  num = num // 10

print(counter)
````

````python
# 12 . Check if all digits are same — 1111 → True, 1121 → False

number = 1111
o_d = number % 10
flag = True          # ✅ assume True first

while number > 0:
    d = number % 10
    if d != o_d:
        flag = False   # ✅ only ever turn OFF, never back on
    number = number // 10

print(flag)
````

````python
num = 44441
o_d = num % 10
counter = True
while num > 0:
  digit = num % 10
  if o_d != digit:
    counter = False

  num = num // 10

print(counter)
````

````python
# 14. Digital Root — Repeatedly sum digits till one digit remains. 9875 → 9+8+7+5=29 → 2+9=11 → 1+1=2
num = 9875

while num > 9:      # Repeat until only one digit remains
    s = 0
    while num > 0:
        d = num % 10
        s += d
        num //= 10
    num = s

print(num)
````
