```python
# 11. Reverse a string using a for loop (no slicing)

original_str = "Python"
reversed_str = ""

for char in original_str:
    # Prepending: Put the new character before the existing ones
    reversed_str = char + reversed_str

print(f"Original: {original_str}")
print(f"Reversed: {reversed_str}")

```

```python
# 12. Count vowels and consonants in a sentence

txt = "Loops are Fun!"
vowel = []
constants = []

for i in txt:
    if i.isalpha():
        if i.lower() in 'aeiou':
            vowel.append(i)
        else:
            constants.append(i.lower())

print(vowel)
print(constants)
```

```python
# 14. Write a program to reverse a given integer number (e.g., 76542 should become 24567).

num = 76542
rev_num = 0

while num > 0:
  digit = num % 10
  rev_num = (rev_num * 10)+ digit
  num = num // 10
print(rev_num)
```

```python
# Write a program to check if a given number is a palindrome. A palindrome number is a number that remains the same when its digits are reversed (e.g., 121, 343).

num = 343
num= str(num)
r_num = num[::-1]

if num == r_num:
  print(f"{num} palidrome number")
else:
  print(f"{num} not palidrome number")
```

```python
num = 343
original = num
rev_num = 0

while num > 0:
    digit = num % 10
    rev_num = (rev_num * 10) + digit
    num = num // 10

if original == rev_num:
    print(f"{original} is a palindrome number")
else:
    print(f"{original} is not a palindrome number")
```

```python
num = 121

o_num = num
r_num = 0

while num > 0:
  digit = num % 10

  r_num = (r_num * 10) + digit

  num = num // 10


if o_num == r_num:
    print(f"{o_num} is a palindrome number")
else:
    print(f"{o_num} is not a palindrome number")
```
