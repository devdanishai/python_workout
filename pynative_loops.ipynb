```python
# 16. Write a program to check if a given number is a palindrome. A palindrome number is a number that remains the same when its digits are reversed (e.g., 121, 343).

num = 343
num= str(num)
r_num = num[::-1]

if num == r_num:
  print(f"{num} palidrome number")
else:
  print(f"{num} not palidrome number")
```

```python
num = 343
original = num
rev_num = 0

while num > 0:
    digit = num % 10
    rev_num = (rev_num * 10) + digit
    num = num // 10

if original == rev_num:
    print(f"{original} is a palindrome number")
else:
    print(f"{original} is not a palindrome number")
```
