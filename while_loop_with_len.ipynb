````python
# 6. Find the sum of all elements in a list using while
listy = [5, 8, 12, 3, 9]
i = 0
sumy = 0
while i < len(listy):
  sumy += listy[i]
  i+=1

print(sumy)
````

````python
# 7. Find the largest number in a list using while
listy = [5, 8, 12, 3, 9]
i = 0
largest = listy[0]
while i < len(listy):
  if listy[i] > largest:
    largest = listy[i]
  i += 1

print(largest)

````

````python
# 8. Find the smallest number in a list using while
listy = [5, 8, 12, 3, 9]
i = 0
smallest = listy[0]
print(smallest)
while i < len(listy):
  if listy[i] < smallest:
    smallest = listy[i]
  i += 1

print(smallest)
````

````python
# 9. Print list elements in reverse order using while (hint: start index from the end)

listy = [5, 8, 12, 3, 9]
i = len(listy) - 1
while i >= 0:      # ✅ keep going until index goes below 0
  print(listy[i])
  i -= 1
````

````python
# 10. Count how many numbers in the list are greater than 10
listy = [5, 8, 12, 3, 9]
i = 0
counter = 0
while i < len(listy):
  if listy[i] > 10:
    counter += 1
  i += 1

print(counter)
````

````python
# 11. Given a list of numbers, print "even number"/"odd number" for each — (your original problem, but now you write it from scratch)
listy = [5, 8, 12, 3, 9]
i = 0
while i < len(listy):
  if listy[i] % 2 == 0:
    print(f"{listy[i]} is even")
  else:
    print(f"{listy[i]} is odd")
  i += 1
````

````python
# 12. Given a list, create a new list containing only even numbers (using while + append)
listy = [5, 8, 12, 3, 9]
i = 0
even = []
while i < len(listy):
  if listy[i] % 2 ==0:
    even.append(listy[i])
  i += 1
print(even)
````

````python
# 13. Search for a target value in a list — print "Found at index X" or "Not found" (this is basically Linear Search)
listy = [5, 8, 12, 3, 9]
target = 121

i = 0
found = False
while i < len(listy):
  if listy[i] == target:
    print(f"found at index {i}")
    found = True
    break
  i += 1
if not found:
  print("target not found")
````

````python
# 14. Given a list, reverse the entire list itself (not just print in reverse — actually build a reversed list)
listy = [5, 8, 12, 3, 9]
r_listy= []

i = len(listy) -1
while i >= 0:
  r_listy.append(listy[i])
  i -= 1
print(r_listy)
````
