### 2. Sum of numbers

Take a number `n` from the user and calculate:

```
1 + 2 + 3 + ... + n
```

````python
def total_all(m):
    n = 1
    total = 0

    while n <= m:
        total += n
        n += 1

    return total


user_input = int(input("Enter num: "))
print(total_all(user_input))
````

### 10. Number guessing with attempts

or

````python
secret = 25
attempts = 5
while True:

  user_input = int(input("enter number"))
  if user_input == secret:
    print("u win")
    break

    attempts -= 1

print("game over")

````

````python
secret = 25
attempts = 5

while attempts > 0:
    user_input = int(input("Enter number: "))

    if user_input == secret:
        print("You win")
        break

    attempts -= 1
    print(f"Wrong! Attempts left: {attempts}")

if attempts == 0:
    print("Game over")
````

### 4. ATM Menu

Starting balance:

```python
balance = 10000
```

Create a menu:

```
1. Deposit
2. Withdraw
3. Check Balance
4. Exit

````python
balance = 10000

while True:
  print(f"\npress 1 for deposit")
  print(f"press 2 for withdraw")
  print(f"press 3 for balance check")
  print(f"press 4 for exit")

  user_input = input("enter your choice: ")

  if user_input == '1':
    amount = int(input("enetr your amoun: "))
    balance += amount
    print("deposit successfully!")

  elif user_input == '2':
    amount = int(input("enetr withdraw amount: "))
    if amount > balance:
      print(" low balance")
    else:
      balance -= amount
      print("withdraw successful")

  elif user_input == '3':
    print(f"your balance is: {balance}")

  elif user_input == '4':
    print("thanks for visiting our bank")
    break
  else:
    print("invalid input")




````

````python
# 8. Shopping cart menu

cart = []

while True:
    print("1. add item")
    print("2. search item")
    print("3. delete item")
    print("4. update item")
    print("5. display cart")
    print("6. exit")

    choice = input("enter ur choice: ")

    if choice == '1':
        item = input("enter item: ")
        cart.append(item)
        print("item added in cart.")

    elif choice == '2':
        item = input("enter item to search: ")
        if item in cart:
            print("item found in cart.")
        else:
            print("item not found in cart.")

    elif choice == '3':
        item = input("enter item to delete: ")
        if item in cart:
            cart.remove(item)
            print("item deleted from cart.")
        else:
            print("item not found in cart.")

    elif choice == '4':
        old_item = input("enter item to update: ")
        if old_item in cart:
            new_item = input("enter new item name: ")
            index = cart.index(old_item)
            cart[index] = new_item
            print("item updated")
        else:
            print("item not found in cart.")

    elif choice == '5':
        print(cart)

    elif choice == '6':
        break

    else:
        print("invalid input")




````
