````python
n = 10
while n > 0:
  print(n)
  n -= 1
print("Blast off!")
````

### 2. Sum of numbers

Take a number `n` from the user and calculate:

```
1 + 2 + 3 + ... + n
```

````python
def total_all(m):
    n = 1
    total = 0

    while n <= m:
        total += n
        n += 1

    return total


user_input = int(input("Enter num: "))
print(total_all(user_input))
````

### 7. Reverse a number

```

```

Output:

```
54321
```

Use a `while` loop.




````python
number = 12345
r_number = 0

while number > 0:
  digit = number % 10
  r_number = (r_number * 10) + digit
  number = number // 10

print(r_number)

````

or

````python
secret = 25
attempts = 5
while True:

  user_input = int(input("enter number"))
  if user_input == secret:
    print("u win")
    break

    attempts -= 1

print("game over")

````

````python
secret = 25
attempts = 5

while attempts > 0:
    user_input = int(input("Enter number: "))

    if user_input == secret:
        print("You win")
        break

    attempts -= 1
    print(f"Wrong! Attempts left: {attempts}")

if attempts == 0:
    print("Game over")
````

### 4. ATM Menu

Starting balance:

```python
balance = 10000
```

Create a menu:

```
1. Deposit
2. Withdraw
3. Check Balance
4. Exit

````python
balance = 10000

while True:
  print(f"\npress 1 for deposit")
  print(f"press 2 for withdraw")
  print(f"press 3 for balance check")
  print(f"press 4 for exit")

  user_input = input("enter your choice: ")

  if user_input == '1':
    amount = int(input("enetr your amoun: "))
    balance += amount
    print("deposit successfully!")

  elif user_input == '2':
    amount = int(input("enetr withdraw amount: "))
    if amount > balance:
      print(" low balance")
    else:
      balance -= amount
      print("withdraw successful")

  elif user_input == '3':
    print(f"your balance is: {balance}")

  elif user_input == '4':
    print("thanks for visiting our bank")
    break
  else:
    print("invalid input")




````

### 8. Shopping cart menu

Create:

```python
cart = []
```

Menu:

```
1. Add item
2. Remove item
3. Show cart
4. Exit
```

````python
cart = []

while True:
  print("\n1. add item")
  print("2. remove item")
  print("3. show cart")
  print("4. exit")
  print("5. serach item")

  choice = input("chose your number: ")

  if choice == '1':
    ask = input("enetr item: ")
    cart.append(ask)
    print("item added into cart")

  elif choice == '2':
    ask = input("enetr item to remove: ")
    cart.remove(ask)
    print("item removed from cart")

  elif choice == '3':
    print(cart)

  elif choice == '4':
    print("thanks for visiting our site")
    break
  elif choice == '5':
    ask = input("enter your item to found in cart")
    if ask in cart:
      print("item found")
    else:
      print("item not found")

  else:
    print("invalid input")



````

````python
cart = []
def add_item():
    ask = input("enetr item: ")
    cart.append(ask)
    print("item added into cart")

def remove_item():
  ask = input("enetr item to remove: ")
  cart.remove(ask)
  print("item removed from cart")

def show_cart():
  print(cart)


def search_item():
  ask = input("enter your item to found in cart")
  if ask in cart:
    print("item found")
  else:
    print("item not found")

while True:
  print("\n1. add item")
  print("2. remove item")
  print("3. show cart")
  print("4. exit")
  print("5. serach item")

  choice = input("chose your number: ")

  if choice == '1':
    add_item()


  elif choice == '2':
    remove_item()

  elif choice == '3':
    show_cart()

  elif choice == '4':
    print("thanks for visiting our site")
    break
  elif choice == '5':
    search_item()

  else:
    print("invalid input")




````
