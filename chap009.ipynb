# class
- Data types are classes, e.g. list, tuple, set, dict, str, int, float
- These are all built-in classes provided by Python
- When you create a variable holding one of these values, that value is an object
- The variable is a name/reference pointing to that object
- object is an instance of class

  e.g. x = [1, 2, 3]
     - → [1, 2, 3] is the object
     - → x refers to it
     - → x is an instance of the list class
     - → type(x) → <class 'list'>

```python
class Car:
    def __init__(self, brand, speed):
        self.brand = brand    # attribute
        self.speed = speed    # attribute

    def drive(self):           # method
        print(f"{self.brand} is driving at {self.speed} km/h.")

    def honk(self):            # method
        print(f"{self.brand} says: Beep beep!")
```

```python
car1 = Car("Toyota", 120)
car2 = Car("Honda", 100)
```

```python
car1.drive()   # Toyota is driving at 120 km/h.
car2.honk()    # Honda says: Beep beep!

print(car1.brand)   # Toyota
print(car2.speed)   # 100
```

```python
class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages
        self.current_page = 0

    def read(self, pages_read):
        self.current_page = self.current_page + pages_read
        print(f"You are now on page {self.current_page} of {self.pages}.")

    def progress(self):
        percent = (self.current_page / self.pages) * 100
        print(f"You have read {percent}% of the book.")
```

```python
book1 = Book("Atomic Habits", 300)
book1.read(45)      # You are now on page 45 of 300.
book1.progress()
book1.read(30)      # You are now on page 75 of 300.
book1.progress()    # You have read 25.0% of the book.
```

```python
class Player:
    def __init__(self, name, country):
        self.name = name
        self.country = country
        self.score = 0          # new attribute — starts at 0
        self.matches_won = 0    # new attribute — starts at 0

    def add_score(self, points):
        self.score = self.score + points
        print(f"{self.name} now has {self.score} points.")

    def win_match(self):
        self.matches_won = self.matches_won + 1
        print(f"{self.name} has won {self.matches_won} matches.")
```

```python
player1 = Player("Messi", "Argentina")
player2 = Player("Ronaldo", "Portugal")

player1.add_score(120)     # Messi now has 120 points.
player1.add_score(50)      # Messi now has 170 points.
player2.win_match()         # Ronaldo has won 1 matches.
```

- Add a method called increment_number_served() that lets you increment

```python
class Restaurant:
  def __init__(self,name,cuisine):
    self.restaurant_name = name
    self.cuisine_type = cuisine
    self.number_served = 0
    # self.number_served = 20

  def describe_restaurant(self):
    print(f"{self.restaurant_name} is serving {self.cuisine_type}")

  def open_restaurant(self):
    print(f"{self.restaurant_name} is open now")

  def set_number_served(self, num):
    self.number_served = num

  def increment_number_served(self, extra):
    self.number_served += extra


restaurant1 = Restaurant('KFC','fast_food')

print(restaurant1.restaurant_name)
print(restaurant1.cuisine_type)
print(restaurant1.number_served)

restaurant1.describe_restaurant()
restaurant1.open_restaurant()
restaurant1.set_number_served(50)
print(restaurant1.number_served)
restaurant1.increment_number_served(20)
print(restaurant1.number_served)
```
