## 1. Strings

```python
word = "python"

# for loop
for ch in word:
    print(ch)

# while loop
n = 0
while n < len(word):
    print(word[n])
    n += 1
```
Output (both):
```
p
y
t
h
o
n
```
A string is just a sequence of characters, so looping over it gives you one character at a time.

## 2. Integers (looping over a range of numbers)

```python
# for loop
for i in range(1, 6):
    print(i)

# while loop
n = 1
while n < 6:
    print(n)
    n += 1
```
Output (both):
```
1
2
3
4
5
```
Numbers themselves aren't "loopable" — you loop over a *range* of them.

## 3. List

```python
fruits = ['apple', 'banana', 'cherry']

# for loop
for item in fruits:

# while loop
n = 0
while n < len(fruits):
    print(fruits[n])
    n += 1
```
Output (both):
```
apple
banana
cherry
```

## 4. Dictionary

Dicts are a bit different since each entry has a **key** and a **value**.

```python
person = {'name': 'Amit', 'age': 25, 'city': 'Delhi'}

# for loop — keys only (default)
for key in person:
    print(key)

# for loop — keys and values together
for key, value in person.items():
    print(key, value)
```
Output:
```
name
age
city
```
```
name Amit
age 25
city Delhi
```

**`while` loop on a dict** is awkward and rarely done, because dicts aren't indexed by position like lists — you can't do `person[0]`. If you really wanted to, you'd have to convert the keys to a list first:

```python
keys = list(person.keys())
n = 0
while n < len(keys):
    print(keys[n], person[keys[n]])
    n += 1
```
Output:
```
name Amit
age 25
city Delhi
```

**Key takeaway for your notes:** `for` loops work naturally on all four types (strings, ranges, lists, dicts). `while` loops work naturally on strings, ranges/numbers, and lists — but for dicts, `for` is really the natural choice, and `while` requires an extra conversion step to even work. That's a good thing to jot down as the practical difference between the two.

# 3 reverse list
For a list, you have several options:

```python
list1 = [1, 2, 3, 4, 5]

# 1. Slicing
print(list1[::-1])

# 2. reversed()
print(list(reversed(list1)))

# 3. .reverse()
list1.reverse()
print(list1)
```

Output:

```text
[5, 4, 3, 2, 1]
```

4. Using a `while` loop:

```python
list1 = [1, 2, 3, 4, 5]
reverse = []
n = len(list1) - 1

while n >= 0:
    reverse.append(list1[n])
    n -= 1

print(reverse)
```

`list.reverse()` changes the original list, while `reversed()` and slicing can give you a reversed version without modifying the original.

5. for loop
```bash
list1 = [1, 2, 3, 4, 5]
reverse = []

for item in list1:
    reverse.insert(0, item)

print(reverse)
```
