**6. Highest Marks**

Find the student with the highest marks **without using `max()`**.

Example output:

```text
Ali 90
```

````python
students = {
    "Ali": 40,
    "Sara": 75,
    "Hamza": 45,
    "Ayesha": 82
}
largest = 0
for v in students.values():
  if v > largest:
    largest = v

print(largest)
````

````python
sentence = "python is fun and python is powerful"
x = sentence.split()
print(x)

wc={}
for i in x:
  if i in wc:
    wc[i] += 1
  else:
    wc[i] = 1

print(wc)
````

**8. Student Management System**

Use a dictionary.

Menu:

```text
1. Add student
2. Search student
3. Update marks
4. Delete student
5. Show all students
6. Exit
```

Use **functions** for every option, just like you did with the shopping cart.

````python
students = {}
while True:
  print("\n1. add student")
  print("2. search student")
  print("3. update student")
  print("4. delete student")
  print("5. show all students")
  print("6. exit")

  choice = input("enter your choice: ")

  if choice == '1':
    name = input("enter student name: ")
    marks = input("enter student marks: ")
    students[name] = marks
    print("student added successfully!.")

  elif choice == '2':
    # ask name for searching student
    name = input("enter name for search: ")
    print(students[name])
    print("student serched successfully!.")

  elif choice == '3':
    #update student
    name = input("enter name: ")
    marks = input("enter updated marks: ")
    students[name] = marks
    print("student updated successfully!.")

  elif choice == '4':
    #delete student
    name = input("enetr name to dlete student")
    del students[name]
    print("student deleted succcessfully")

  elif choice == '5':
    # show all students
    print(students)

  elif choice == '6':
    break

  else:
    print("invalid input")






````

````python
students = {}
def add_student():
  name = input("enter student name: ")
  marks = int(input("enter student marks: "))
  students[name] = marks
  print("student added successfully!.")

def search_student():

  name = input("enter name for search: ")
  if name in students:
    print(students[name])
    print("student serched successfully!.")
  else:
    print("student not in data")

def update_student():
  name = input("enter name: ")
  marks = int(input("enter updated marks: "))
  students[name] = marks
  print("student updated successfully!.")

def delete_student():
  name = input("enetr name to dlete student")
  del students[name]
  print("student deleted succcessfully")

while True:
  print("\n1. add student")
  print("2. search student")
  print("3. update student")
  print("4. delete student")
  print("5. show all students")
  print("6. exit")

  choice = input("enter your choice: ")

  if choice == '1':
    add_student()

  elif choice == '2':
    search_student()


  elif choice == '3':
    update_student()


  elif choice == '4':
    delete_student()

  elif choice == '5':
    # show all students
    print(students)

  elif choice == '6':
    break

  else:
    print("invalid input")
````

**9. Phone Book**

Use a dictionary:

Menu:

````python
ph_book = {}

while True:
  print("\n1. add contact")
  print("2. search contact")
  print("3. Update number")
  print("4. delete contact")
  print("5. show all contacts")
  print("6. exit")

  choice = input("enter ur choice: ")

  if choice == '1':
    name = input("enter name: ")
    number = input("enter number: ")
    ph_book[name] = number
    print("number added successfully!")

  elif choice == '2':
    #search
    name = input("enter name: ")
    if name in ph_book:
      print(ph_book[name])
    else:
      print("name is not in phone book")

  elif choice == '3':
    #update
    name = input("enter name: ")
    number = input("enter new number: ")
    if name in ph_book:
      ph_book[name] = number
      print("number updated!")
    else:
      print("name is not found")

  elif choice == '4':
    # delete
    name = input("enter name for delete: ")
    if name in ph_book:
      del ph_book[name]
      print("number deleted!")
    else:
      print("name noty found")

  elif choice == '5':
    # show all
    print(ph_book)

  elif choice == '6':
    break

  else:
    print("invalid number")
````
